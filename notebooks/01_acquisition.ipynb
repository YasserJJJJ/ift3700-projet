# Tâche 1 — Acquisition

Ce notebook ne contient pas de résultats préexécutés. Les commandes doivent télécharger de vrais matchs depuis une machine ayant accès à l’API.


```python
import os
from pathlib import Path
import sys
PROJECT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT))
DATA_DIR = Path(os.environ.get('NHL_DATA_DIR') or PROJECT / 'data/raw')
```

```python
from ift3700 import NHLClient
client = NHLClient(DATA_DIR)
game = client.get_game(2017020001)
print(game['id'], len(game['plays']), 'événements')
```

Le deuxième appel doit utiliser le cache et ne faire aucune nouvelle requête HTTP.


```python
before = client.request_count
same_game = client.get_game(2017020001)
assert same_game == game
assert client.request_count == before
print('Cache vérifié ; compteur de requêtes :', client.request_count)
```

La cellule suivante télécharge les huit saisons complètes, saison régulière et séries. Elle peut être longue ; les fichiers déjà présents seront chargés localement lors d’une relance.


```python
reports = []
for year in range(2016, 2024):
    report = client.download_season(year)
    reports.append(report)
    print(report['season'], 'complet :', report['complete'], 'échecs :', len(report['failed']))
assert all(report['complete'] for report in reports), 'Relancer et examiner les rapports des échecs.'
```
